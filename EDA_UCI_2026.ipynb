```python
from ucimlrepo import fetch_ucirepo

# fetch dataset
heart_disease = fetch_ucirepo(id=45)

# data (as pandas dataframes)
X = heart_disease.data.features
y = heart_disease.data.targets

# metadata
print(heart_disease.metadata)

# variable information
print(heart_disease.variables)
```

```python
import pandas as pd

df = pd.DataFrame(X, columns=heart_disease.name)
df['target'] = y

df.head()
```

### Descripción de las Columnas (Atributos)

Basado en la información de las variables del conjunto de datos:

- **age**: Edad del paciente en años.
- **sex**: Sexo del paciente (1 = masculino; 0 = femenino).
- **cp**: Tipo de dolor en el pecho.
  - Valor 1: angina típica
  - Valor 2: angina atípica
  - Valor 3: dolor no anginoso
  - Valor 4: asintomático
- **trestbps**: Presión arterial en reposo (en mm Hg al ingreso al hospital).
- **chol**: Colesterol sérico en mg/dl.
- **fbs**: Azúcar en sangre en ayunas > 120 mg/dl (1 = verdadero; 0 = falso).
- **restecg**: Resultados electrocardiográficos en reposo.
  - Valor 0: normal
  - Valor 1: anomalía de la onda ST-T (inversión de la onda T y/o elevación o depresión del ST > 0.05 mV)
  - Valor 2: hipertrofia ventricular izquierda probable o definida según los criterios de Estes
- **thalach**: Frecuencia cardíaca máxima alcanzada.
- **exang**: Angina inducida por el ejercicio (1 = sí; 0 = no).
- **oldpeak**: Depresión del ST inducida por el ejercicio en relación con el reposo.
- **slope**: La pendiente del segmento ST máximo del ejercicio.
  - Valor 1: ascendente
  - Valor 2: plano
  - Valor 3: descendente
- **ca**: Número de vasos principales coloreados por fluoroscopia (0-3).
- **thal**: Tipo de talasemia.
  - Valor 3: normal
  - Valor 6: defecto fijo
  - Valor 7: defecto reversible
- **target**: El atributo predicho. Indica la presencia de enfermedad cardíaca.
  - Valor 0: ausencia de enfermedad cardíaca
  - Valores 1, 2, 3, 4: presencia de enfermedad cardíaca (cuanto mayor el valor, mayor la severidad)

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Configuramos el estilo del gráfico
sns.set_theme(style='whitegrid')

# Creamos el gráfico de frecuencias para la columna 'thal'
plt.figure(figsize=(8, 6))
# Corregimos la advertencia asignando x a hue y desactivando la leyenda
ax = sns.countplot(data=df, x='thal', hue='thal', palette='viridis', legend=False)

# Añadimos títulos y etiquetas
plt.title('Frecuencia de Tipos de thal')
plt.xlabel('Tipo de Thal')
plt.ylabel('Frecuencia')

# Mostramos el gráfico
plt.show()
```
